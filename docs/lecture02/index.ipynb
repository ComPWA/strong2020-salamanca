## Experiments and frames

Particle kinematics is the application of special relativity to elementary particle reactions.
In experiments, the "beam" collides onto the "target" with high energy to create other particles.

Each particle has a 4-vector $p^{\mu}$:

$$
p^{\mu} = (E, p_x, p_y, p_z) = (E, \vec{p})
$$

Two frames that are commonly used are the Center of Mass (CoM) frame and the laboratory (Lab) frame.

For a two-particle reaction, $a+b \to 1 + ... + n$, the frames satisfy the following relations.
- **Lab frame**: $\vec{p}_b = \vec{0}$<br>
  Target particle $b$ is at rest, while beam particle $a$ is moving.
- **CoM frame**: $\vec{p}_a+\vec{p}_b = \vec{0}$<br>
  The total momentum is zero, i.e., the momenta of the two initial state particles have opposite direction.

Masses can be defined as a Lorentz invariant quantity (unaffected by reference frame):

$$
p^2_i = E^2_i - |\vec{p}_i|^2 = m^2_i,
$$

where $p$ is the 4-vector, $E$ is the energy, $\vec{p}$ is the three vector, and $m$ is the mass.

When we have a final state with two particles ($n=2$), we can define one $xz$ plane that is spanned by the momenta of the two final state particles in the Lab frame. The CoM frame and the Lab frame can now be related by a boost along the $z$ axis (the direction of particle $a$).

With this definition of the $xz$ plane, we have:
- $\vec{z}$ is parallel to the beam $\vec{p}_a$
- $\vec{y}$ is parallel to $\vec{p}_a \times \vec{p}_1$

## Lorentz transformations

Lorentz transformations can be factorized into rotations and boosts along one axis and are required for transforming particle properties, such as position and time, between different inertial frames in relative motion.

### Boost along one axis

When we boost in the direction of the $z$ axis, only the energy and the $z$ component of the four-momenta changes, so we can express the Lorentz boost as a 2D matrix:

$$\begin{pmatrix}
 E^*
 \\
 p_z^*
\end{pmatrix} =
\begin{pmatrix}
  \gamma & \gamma \beta \\
  \gamma \beta & \gamma
\end{pmatrix}
\begin{pmatrix}
 E^L\\
 p^L_z
\end{pmatrix},
$$

and its inverse relation,

$$\begin{pmatrix}
 E^L
 \\
 p_z^L
\end{pmatrix} =
\begin{pmatrix}
  \gamma & -\gamma \beta \\
  -\gamma \beta & \gamma
\end{pmatrix}
\begin{pmatrix}
 E^*\\
 p^*_z
\end{pmatrix},
$$ (inverse_relation)

with

$$
\beta = \frac{p^*_{b,z}}{E^*_b}
$$ (beta_label)

and

$$
\gamma = \frac{1}{\sqrt{1-\beta^2}} = \frac{E^*_b}{m_b},
$$ (gamma_label)

:::

:::{solution} boost-exercise
:class: dropdown

We want to show that the boost {eq}`inverse_relation` from the CoM frame to the lab frame brings the target $b$ to rest, that is,

$$
\begin{pmatrix}
 E_b^L \\
 p_{b,z}^L
\end{pmatrix} =
\begin{pmatrix}
 m_b \\
 0
\end{pmatrix}.
$$

In the CoM frame, the target moves along the $z$ axis, so $p_{b,x}^* = p_{b,y}^* = 0$. A boost along the $z$ axis does not change these components, so they also vanish in the lab frame and we only have to consider the energy and the $z$ component.

We first express $\gamma$ and $\gamma\beta$ in terms of the CoM energy and momentum of the target. Inserting the velocity {eq}`beta_label` into the definition of $\gamma$ and using the mass-shell relation $E_b^{*2} - p_{b,z}^{*2} = m_b^2$ gives

$$
\gamma = \frac{1}{\sqrt{1-\beta^2}} = \frac{E_b^*}{\sqrt{E_b^{*2} - p_{b,z}^{*2}}} = \frac{E_b^*}{m_b},
\qquad
\gamma\beta = \frac{E_b^*}{m_b} \cdot \frac{p_{b,z}^*}{E_b^*} = \frac{p_{b,z}^*}{m_b},
$$

which is the result in {eq}`gamma_label`. Substituting these into {eq}`inverse_relation` for particle $b$, we get

$$
\begin{pmatrix}
 E_b^L \\
 p_{b,z}^L
\end{pmatrix} =
\frac{1}{m_b}
\begin{pmatrix}
  E_b^* & -p_{b,z}^* \\
  -p_{b,z}^* & E_b^*
\end{pmatrix}
\begin{pmatrix}
 E_b^* \\
 p_{b,z}^*
\end{pmatrix} =
\frac{1}{m_b}
\begin{pmatrix}
 E_b^{*2} - p_{b,z}^{*2} \\
 -p_{b,z}^* E_b^* + E_b^* p_{b,z}^*
\end{pmatrix} =
\begin{pmatrix}
 m_b \\
 0
\end{pmatrix}.
$$

The target therefore has no momentum in the lab frame and its energy equals its mass, so it is at rest.
:::

### Rotations

We use _active_ rotations, meaning that the momentum itself is rotated while the coordinate axes remain fixed.

As an example, take a momentum of unit length in the $xz$ plane that forms an angle $\theta$ with the $z$ axis:

$$
\begin{pmatrix}
 p_x
 \\
 p_y
 \\
 p_z
\end{pmatrix} =
\begin{pmatrix}
 \sin(\theta)
 \\
 0
 \\
 \cos(\theta)
\end{pmatrix}.
$$

After a rotation over an angle $\omega$ around the $y$ axis, the rotated momentum forms an angle $\theta + \omega$ with the $z$ axis:

$$
\begin{pmatrix}
 p_x'
 \\
 p_y'
 \\
 p_z'
\end{pmatrix}
=
R_y(\omega)
\begin{pmatrix}
 p_x
 \\
 p_y
 \\
 p_z
\end{pmatrix}
=
\begin{pmatrix}
 \sin(\theta+\omega)
 \\
 0
 \\
 \cos(\theta+\omega)
\end{pmatrix}.
$$ (rotation_label)

:::{exercise}
:label: check-rotation
Check the result of the rotation above.
:::

:::{solution} check-rotation
:class: dropdown

Recall the rotation matrices. Any rotation can be decomposed into rotations around the $z$ and $y$ axes:

$$
R_z(\omega)
=
\begin{pmatrix}
\cos(\omega) & -\sin(\omega) & 0 \\
\sin(\omega) & \cos(\omega) & 0 \\
0 & 0 & 1
\end{pmatrix},
$$

$$
R_y(\omega)
=
\begin{pmatrix}
\cos(\omega) & 0 & \sin(\omega) \\
0 & 1 & 0 \\
-\sin(\omega) &0 & \cos(\omega)
\end{pmatrix}.
$$

The rotation matrix around the $x$ axis is also given for completeness:

$$
R_x(\omega)
=
\begin{pmatrix}
1&0&0 \\
0 & \cos(\omega) & -\sin(\omega) \\
0 & \sin(\omega) & \cos(\omega)
\end{pmatrix}.
$$

With this, we can rewrite the rotation of {eq}`rotation_label` as

$$
\begin{pmatrix}
 p_x' \\
 p_y' \\
 p_z'
\end{pmatrix}
=
R_y(\omega)
\begin{pmatrix}
 p_x \\
 p_y \\
 p_z
\end{pmatrix}
=
\begin{pmatrix}
\cos(\omega) & 0 & \sin(\omega) \\
0 & 1 & 0 \\
-\sin(\omega) &0 & \cos(\omega)
\end{pmatrix}
\begin{pmatrix}
 p_x \\
 p_y \\
 p_z
\end{pmatrix}
=
\begin{pmatrix}
p_x \cdot \cos(\omega)+ p_z \cdot \sin(\omega) \\
p_y \\
-p_x \cdot \sin(\omega) + p_z \cdot \cos(\omega)
\end{pmatrix}.
$$

Since $(p_x, p_y, p_z) = (\sin(\theta), 0, \cos(\theta))$, and using the trigonometric identities $\sin(\theta+\omega) = \sin(\theta)\cos(\omega) + \cos(\theta)\sin(\omega)$ and $\cos(\theta+\omega) = \cos(\theta)\cos(\omega) - \sin(\theta)\sin(\omega)$,
we get:

$$
\begin{pmatrix}
 p_x' \\
 p_y' \\
 p_z'
\end{pmatrix}
=
\begin{pmatrix}
\sin(\theta) \cdot \cos(\omega)+ \cos(\theta) \cdot \sin(\omega) \\
0 \\
-\sin(\theta) \cdot \sin(\omega) + \cos(\theta) \cdot \cos(\omega)
\end{pmatrix}
=
\begin{pmatrix}
\sin(\theta+\omega) \\
0 \\
\cos(\theta+\omega)
\end{pmatrix}.
$$
:::

We first download the data file with [`gdown`](https://github.com/wkentaro/gdown) and load it with NumPy.

````python
filename = gdown.cached_download(
    url="https://indico.ific.uv.es/event/6803/contributions/21220/attachments/11209/15506/Two-particles-1.dat",
    path="data/Two-particles-1.dat",
    hash="md5:324f71ee88928915f13de5840ac5ede2",
    quiet=True,
    verify=False,
)
data = np.loadtxt(filename)
data.shape
````

The data file lists the four-momenta $p_a$, $p_1$, and $p_2$ of the reaction $a+b \to 1+2$ for each event, one after the other. We split them into three arrays.

````python
pa = data[::3].T
p1 = data[1::3].T
p2 = data[2::3].T
pa.shape
````

The four-momentum of the target $p_b$ follows from four-momentum conservation, $p_a + p_b = p_1 + p_2$. We check that it is the same for all events.

````python
pb = p1 + p2 - pa
assert pb.std(axis=1).sum().round(5) == 0
pb.mean(axis=1).round(3)
````

The mass of particle&nbsp;1 is given by $m_1 = \sqrt{E_1^2 - |\vec{p}_1|^2}$:

````python
m1 = np.sqrt(p1[0] ** 2 - p1[1] ** 2 - p1[2] ** 2 - p1[3] ** 2)
m1.mean(), m1.std()
````

The scattering angle $\theta$ is the angle between the momentum of particle&nbsp;1 and the beam axis $z$, so

$$
\cos \theta = \frac{p_{1z}}{|\vec{p_1}|}.
$$ (cos_theta_label)

````python
def p3_length(four_momentum: np.ndarray) -> np.ndarray:
    return np.sqrt(np.sum(four_momentum[1:] ** 2, axis=0))


p1_length = p3_length(p1)
cos_theta = p1[3] / p1_length
````

````python
import matplotlib.pyplot as plt

plt.title(r"$\cos \theta$ distribution in lab frame")
plt.xlabel(r"$\cos \theta$")
plt.hist(cos_theta, bins=40)
plt.show()
````

#### $\cos\theta$ in the CoM frame

:::{exercise}
:label: two-particles-cm
Boost the four-momenta of {ref}`two-particles-lab` to the CoM frame and compute and plot the $\cos\theta^*$ distribution.
:::

:::{solution-start} two-particles-cm
:class: dropdown
:::

We first perform the boost by hand with NumPy.

The CoM frame moves with velocity $\vec{\beta} = \vec{P}/E$ with respect to the lab frame, where $P = p_a + p_b = (E, \vec{P})$ is the total four-momentum of the initial state. Since the target is at rest in the lab frame, $\vec{P} = \vec{p}_a$ and $E = E_a + m_b$.

````python
P = pa + pb
β_CM = P[1:] / P[0]
β2 = np.sum(β_CM**2, axis=0)
gamma = 1 / np.sqrt(1 - β2)
````

The energy of particle&nbsp;1 in the CoM frame follows from the Lorentz boost

$$
E_1^* = \gamma\left(E_1 - \vec{\beta} \cdot \vec{p}_1\right),
$$

with $\gamma = 1/\sqrt{1-\beta^2}$.

````python
β_dot_p1 = np.sum(β_CM * p1[1:], axis=0)
E1_CM = gamma * (p1[0] - β_dot_p1)
````

Its three-momentum in the CoM frame is

$$
\vec{p}_1^{\,*} = \vec{p}_1 + \left(\frac{\gamma - 1}{\beta^2} \left(\vec{\beta} \cdot \vec{p}_1\right) - \gamma E_1\right) \vec{\beta}.
$$

````python
p1_CM_vec = p1[1:] + ((gamma - 1) / β2 * β_dot_p1 - gamma * p1[0]) * β_CM
p1_CM = np.concatenate((E1_CM[None], p1_CM_vec))
````

Equation {eq}`cos_theta_label` also holds in the CoM frame, so we find $\cos\theta^*$ with:

````python
p1_CM_length = p3_length(p1_CM)
cos_theta_CM = p1_CM[3] / p1_CM_length
````

````python
plt.hist(cos_theta_CM, bins=40)
plt.title(r"$\cos \theta$ distribution in CoM frame")
plt.xlabel(r"$\cos \theta$")
plt.show()
````

We check the boost with the [`pylorentz`](https://pypi.org/project/pylorentz) package and compare the $\cos\theta^*$ values computed in both ways. The method `boost_particle()` transforms a four-vector from the rest frame of the given momentum into the lab frame. To transform from the lab frame into the CoM frame, we therefore pass it the total four-momentum with reversed three-momentum.

````python
from pylorentz import Momentum4

p1_Lab = Momentum4(*p1)
pa_Lab = Momentum4(*pa)
pb_Lab = Momentum4(*pb)

cm = pa_Lab + pb_Lab
cm_reversed = Momentum4(cm.e, -cm.p_x, -cm.p_y, -cm.p_z)
p1_CM_pyL = p1_Lab.boost_particle(cm_reversed)

cos_theta_CM_pyL = p1_CM_pyL.p_z / p1_CM_pyL.p

np.testing.assert_allclose(cos_theta_CM, cos_theta_CM_pyL, rtol=1e-5)
print("Test passed with relative tolerance 1e-5")
````

## Mandelstam variables

The Mandelstam variables $s$, $t$, and $u$ are Lorentz invariants, defined as

$$
\begin{aligned}
s &= (p_a + p_b)^2 = (p_1+p_2)^2, \\
t &= (p_a - p_1)^2 = (p_b-p_2)^2, \\
u &= (p_a - p_2)^2 = (p_b-p_1)^2.
\end{aligned}
$$


```{image} https://github.com/ComPWA/strong2020-salamanca/assets/17490173/4daa6f59-615d-4950-b7ba-e8a93f04253c
:width: 300px
```

Using the mass-shell relation $p_i^2 = m_i^2$, their sum only depends on the masses of the particles:

$$
s+t+u = m_a^2+m_b^2+m_1^2+m^2_2.
$$ (mandelstam_sum)

:::{exercise}
:label: mv
Check the relation in Equation {eq}`mandelstam_sum` by using $p_a+p_b=p_1+p_2$.
:::

:::{solution} mv
:class: dropdown

Expanding the squares in the definitions of $s$, $t$, and $u$ gives

$$
s+t+u = 3p_a^2 + p_b^2 + p_1^2 + p_2^2 + 2p_a \cdot p_b - 2p_a \cdot p_1 - 2p_a \cdot p_2.
$$

Four-momentum conservation, $p_b = p_1 + p_2 - p_a$, gives

$$
p_a \cdot p_b = p_a \cdot p_1 + p_a \cdot p_2 - p_a^2.
$$

Substituting this into the expression above, all mixed products cancel and we are left with

$$
s+t+u = p_a^2 + p_b^2 + p_1^2 + p_2^2 = m_a^2 + m_b^2 + m_1^2 + m_2^2,
$$

where we used the mass-shell relation in the last step.
:::

### CoM kinematics

Every frame-dependent quantity can be expressed in terms of the Mandelstam variables. In the CoM frame, the four-momenta of the reaction $a+b \to 1+2$ are


```{image} https://user-images.githubusercontent.com/17490173/278340815-e5eb448f-e620-4b67-873a-e18cd146c55a.png
:width: 500px
```

$$
\begin{aligned}
p_a &= \left(E_a^*, 0, 0, |\vec{p}^*_a|\right), \\
p_b &= \left(E_b^*, 0, 0, -|\vec{p}^*_a|\right), \\
p_1 &= \left(E_1^*, |\vec{p}^*_1| \sin\theta^*, 0, |\vec{p}^*_1| \cos\theta^*\right), \\
p_2 &= \left(E_2^*, -|\vec{p}^*_1| \sin\theta^*, 0, -|\vec{p}^*_1| \cos\theta^*\right),
\end{aligned}
$$

so that

$$
s = (p_a + p_b)^2 = (E_a^* + E_b^*)^2 = E_{CM}^2
$$

and

$$
t = (p_a - p_1)^2 = m_a^2 + m_1^2 - 2E_a^* E_1^* + 2 |\vec{p}^*_a| |\vec{p}^*_1| \cos\theta^*.
$$ (t_cm)

:::{exercise}
:label: E2*
Check that $E_1^* = \frac{s+m^2_1-m_2^2}{2 \sqrt{s}}$ and $E_2^* = \frac{s+m^2_2-m_1^2}{2 \sqrt{s}}$. Hint: square $p_2 = (p_a+p_b) - p_1$.
:::

:::{solution} E2*
:class: dropdown

Write $P = p_a + p_b$ for the total four-momentum. In the CoM frame, $P = (\sqrt{s}, \vec{0})$, so that $P \cdot p_1 = \sqrt{s}\,E_1^*$. Squaring $p_2 = P - p_1$ then gives

$$
m_2^2 = p_2^2 = P^2 - 2P \cdot p_1 + p_1^2 = s - 2\sqrt{s}\,E_1^* + m_1^2.
$$

Solving for $E_1^*$ gives

$$
E_1^* = \frac{s + m_1^2 - m_2^2}{2\sqrt{s}}.
$$

Energy conservation in the CoM frame, $E_1^* + E_2^* = \sqrt{s}$, then gives

$$
E_2^* = \sqrt{s} - E_1^* = \frac{s + m_2^2 - m_1^2}{2\sqrt{s}}.
$$
:::

:::{exercise}
:label: p1*
Check that $|\vec{p}^*_1| = \sqrt{E_1^{*2}-m_1^2} = \frac{\lambda^{1/2}(s,m_1^2,m_2^2)}{2 \sqrt{s}}$, where $\lambda(x,y,z) = x^2+y^2+z^2 - 2(xy+yz+zx)$ is the Källén function.
:::

:::{solution} p1*
:class: dropdown

Inserting $E_1^*$ from {ref}`E2*` gives

$$
|\vec{p}^*_1|^2 = E_1^{*2} - m_1^2 = \frac{(s + m_1^2 - m_2^2)^2 - 4sm_1^2}{4s}.
$$

Expanding the numerator,

$$
(s + m_1^2 - m_2^2)^2 - 4sm_1^2 = s^2 + m_1^4 + m_2^4 - 2sm_1^2 - 2sm_2^2 - 2m_1^2m_2^2 = \lambda(s, m_1^2, m_2^2),
$$

so that $|\vec{p}^*_1| = \lambda^{1/2}(s,m_1^2,m_2^2) / 2\sqrt{s}$. The same steps for the initial state give $|\vec{p}^*_a| = \lambda^{1/2}(s,m_a^2,m_b^2) / 2\sqrt{s}$.
:::

Since $-1 \leq \cos\theta^* \leq 1$, Equation {eq}`t_cm` shows that $t$ lies between the values for forward ($\theta^* = 0$) and backward ($\theta^* = \pi$) scattering,

$$
t_{0,\pi} = m_a^2 + m_1^2 - 2E_a^* E_1^* \pm 2 |\vec{p}^*_a| |\vec{p}^*_1|.
$$

The forward value $t_0$ has the smallest magnitude $|t|$ and is therefore usually denoted $t_\mathrm{min}$.

With this, the scattering angle in the CoM frame can be expressed in terms of the Mandelstam variables as

$$
\cos\theta^* = 1 + \frac{t-t_\mathrm{min}}{2|\vec{p}^*_a||\vec{p}^*_1|}
=
\frac{s(t-u)+(m_a^2-m_b^2)(m_1^2-m_2^2)}{\lambda^{1/2}(s,m_a^2,m_b^2)\,\lambda^{1/2}(s,m_1^2,m_2^2)}.
$$ (cos_theta_cm_stu)

:::{exercise}
:label: cos-theta-lab
Compute $\cos\theta^L$, the scattering angle of particle&nbsp;1 in the lab frame, as a function of $s$, $t$, and $u$.
:::

:::{solution-start} cos-theta-lab
:class: dropdown
:::

In the lab frame, the target is at rest, $p_b = (m_b, \vec{0})$. The energies of particles&nbsp;$a$ and&nbsp;1 then follow from

$$
\begin{aligned}
s &= (p_a + p_b)^2 = m_a^2 + m_b^2 + 2m_b E_a^L, \\
u &= (p_b - p_1)^2 = m_b^2 + m_1^2 - 2m_b E_1^L,
\end{aligned}
$$

and their momenta from the mass-shell relation, in the same way as in {ref}`p1*`:

$$
\begin{aligned}
E_a^L &= \frac{s - m_a^2 - m_b^2}{2m_b}, &
|\vec{p}^L_a| &= \frac{\lambda^{1/2}(s, m_a^2, m_b^2)}{2m_b}, \\
E_1^L &= \frac{m_b^2 + m_1^2 - u}{2m_b}, &
|\vec{p}^L_1| &= \frac{\lambda^{1/2}(u, m_b^2, m_1^2)}{2m_b}.
\end{aligned}
$$

Just like Equation {eq}`t_cm` in the CoM frame, we have $t = m_a^2 + m_1^2 - 2E_a^L E_1^L + 2|\vec{p}^L_a||\vec{p}^L_1|\cos\theta^L$ in the lab frame. Solving for $\cos\theta^L$ gives

$$
\cos\theta^L = \frac{2m_b^2\left(t - m_a^2 - m_1^2\right) + \left(s - m_a^2 - m_b^2\right)\left(m_b^2 + m_1^2 - u\right)}{\lambda^{1/2}(s, m_a^2, m_b^2)\,\lambda^{1/2}(u, m_b^2, m_1^2)}.
$$ (cos_theta_lab_stu)

We check Equations {eq}`cos_theta_cm_stu` and {eq}`cos_theta_lab_stu` with the four-momenta of {ref}`two-particles-lab` and compare them with the angles that we computed directly from the four-momenta. The tolerance accounts for the limited number of digits in the data file.

````python
def minkowski_square(p: np.ndarray) -> np.ndarray:
    return p[0] ** 2 - np.sum(p[1:] ** 2, axis=0)


def kallen(x: np.ndarray, y: np.ndarray, z: np.ndarray) -> np.ndarray:
    return x**2 + y**2 + z**2 - 2 * (x * y + y * z + z * x)


s = minkowski_square(pa + pb)
t = minkowski_square(pa - p1)
u = minkowski_square(pa - p2)
ma_sq, mb_sq, m1_sq, m2_sq = (minkowski_square(p) for p in (pa, pb, p1, p2))

cos_theta_CM_stu = (s * (t - u) + (ma_sq - mb_sq) * (m1_sq - m2_sq)) / np.sqrt(
    kallen(s, ma_sq, mb_sq) * kallen(s, m1_sq, m2_sq)
)
cos_theta_lab_stu = (
    2 * mb_sq * (t - ma_sq - m1_sq) + (s - ma_sq - mb_sq) * (mb_sq + m1_sq - u)
) / np.sqrt(kallen(s, ma_sq, mb_sq) * kallen(u, mb_sq, m1_sq))

np.testing.assert_allclose(cos_theta_CM_stu, cos_theta_CM, atol=1e-5)
np.testing.assert_allclose(cos_theta_lab_stu, cos_theta, atol=1e-5)
````

## Three-particle final states

For a reaction $a+b \to 1+2+3$, we describe the final state with the invariant masses squared of the three pairs of final-state particles, $s_{ij} = (p_i + p_j)^2$. Only two of them are independent, because their sum is fixed by the invariant mass squared $M^2 = (p_1+p_2+p_3)^2 = s$ of the final state and the masses of the final-state particles:

$$
s_{12} + s_{23} + s_{31} = M^2 + m_1^2 + m_2^2 + m_3^2.
$$ (dalitz_sum)

:::{exercise}
:label: dalitz-sum
Check the relation in Equation {eq}`dalitz_sum`.
:::

:::{solution} dalitz-sum
:class: dropdown

Write $P = p_1 + p_2 + p_3$, so that $P^2 = M^2$. Then $s_{12} = (P - p_3)^2 = M^2 - 2P \cdot p_3 + m_3^2$ and similarly for $s_{23}$ and $s_{31}$. Summing the three gives

$$
s_{12} + s_{23} + s_{31} = 3M^2 - 2P \cdot (p_1 + p_2 + p_3) + m_1^2 + m_2^2 + m_3^2 = M^2 + m_1^2 + m_2^2 + m_3^2.
$$
:::

## Dalitz plot analysis

We load the data and split it into the four-momenta of the particles. The four-momentum of the target&nbsp;$b$ follows from four-momentum conservation.

````python
filename = gdown.cached_download(
    url="https://indico.ific.uv.es/event/6803/contributions/21220/attachments/11209/15504/Three-particles-1.dat",
    path="data/Three-particles-1.dat",
    hash="md5:a49ebfd97ae6a02023291df665ab924c",
    quiet=True,
    verify=False,
)
data = np.loadtxt(filename)
pa, p1, p2, p3 = (data[i::4].T for i in range(4))
p0 = p1 + p2 + p3
pb = p0 - pa
````

The four-momentum of the target is opposite to that of the beam, which confirms that the data are given in the CoM frame:

````python
pa.mean(axis=1).round(3), pb.mean(axis=1).round(3)
````

The masses of the particles identify the reaction. Here, $m_0$ is the invariant mass of the final state, that is, the CoM energy $\sqrt{s}$.

````python
def mass(p: np.ndarray) -> np.ndarray:
    return np.sqrt(mass_squared(p))


def mass_squared(p: np.ndarray) -> np.ndarray:
    return p[0] ** 2 - np.sum(p[1:] ** 2, axis=0)
````

````python
from IPython.display import Math
````

````python
masses = {
    "a": mass(pa).mean(),
    "b": mass(pb).mean(),
    **{i: mass(p).mean() for i, p in enumerate([p0, p1, p2, p3])},
}
rows = R" \\ ".join(
    Rf"m_{{{label}}} &= {value:.3g}\text{{ GeV}}" for label, value in masses.items()
)
Math(Rf"""
\begin{{aligned}}
{rows}
\end{{aligned}}
""")
````

We look up particles with these masses with the [`particle`](https://github.com/scikit-hep/particle) package. The gluon is excluded, because it is massless like the photon.

````python
from particle import Particle


def find_candidates(
    m: float, delta: float = 0.001, charge: float | None = None
) -> list[Particle]:
    def identify(p: Particle) -> bool:
        if p.pdgid == 21:
            return False
        if charge is not None and p.charge != charge:
            return False
        return m - delta < 1e-3 * p.mass < m + delta

    return Particle.findall(identify)
````

````python
initial_state = (
    find_candidates(mass(pa).mean(), delta=1e-4)[0],
    find_candidates(mass(pb).mean())[0],
)
final_state = tuple(find_candidates(mass(p).mean())[0] for p in (p1, p2, p3))
incoming = ", ".join(p.latex_name for p in initial_state)
outgoing = ", ".join(p.latex_name for p in final_state)
Math(Rf"""
\begin{{aligned}}
\text{{Incoming: }} & {incoming} \\
\text{{Outgoing: }} & {outgoing}
\end{{aligned}}
""")
````

So this is a photon&nbsp;$\gamma$ hitting a proton&nbsp;$p$ at $\sqrt{s} \approx 4.1\text{ GeV}$, producing an $\eta$&nbsp;meson, a pion&nbsp;$\pi^0$, and a proton&nbsp;$p$.

A Dalitz plot shows two of the three invariant masses squared $s_{ij}$ against each other. Equation {eq}`dalitz_sum` shows that the third one carries no additional information. We check this relation with the data:

````python
np.testing.assert_allclose(
    mass_squared(p1 + p2) + mass_squared(p2 + p3) + mass_squared(p3 + p1),
    mass_squared(p0) + mass_squared(p1) + mass_squared(p2) + mass_squared(p3),
    rtol=1e-5,
)
````

A resonance in the $ij$ subsystem shows up as a band of constant $s_{ij}$ in the Dalitz plot. We estimate the positions $R_{ij}$ of these bands from the peaks in the distributions of $s_{12}$, $s_{23}$, and $s_{31}$.

````python
import matplotlib.pyplot as plt

COLORS = {"12": "C0", "23": "C1", "31": "C2"}


def find_peak(s: np.ndarray, bins: int = 200) -> float:
    counts, edges = np.histogram(s, bins=bins)
    i = counts.argmax()
    return (edges[i] + edges[i + 1]) / 2


def axis_label(ij: str) -> str:
    return Rf"$s_{{{ij}}}\;\left[\mathrm{{GeV}}^2\right]$"


def plot_dalitz(s: dict[str, np.ndarray], peaks: dict[str, float]) -> None:
    fig, (ax1, ax2, ax3) = plt.subplots(figsize=(9.5, 3.2), ncols=3)
    fig.suptitle("Dalitz plot")
    ax1.hist2d(s["12"], s["23"], bins=100, cmin=1)
    ax1.set_xlabel(axis_label("12"))
    ax1.set_ylabel(axis_label("23"))
    for ax, x, y in [(ax2, "12", "23"), (ax3, "31", "12")]:
        ax.scatter(s[x], s[y], c="black", s=1e-3, rasterized=True)
        ax.set_xlabel(axis_label(x))
        ax.set_ylabel(axis_label(y))
        ax.axvline(peaks[x], c=COLORS[x], ls="dashed", label=f"$R_{{{x}}}$")
        ax.axhline(peaks[y], c=COLORS[y], ls="dashed", label=f"$R_{{{y}}}$")
        ax.legend()
    fig.tight_layout()
    plt.show()


def plot_projections(s: dict[str, np.ndarray], peaks: dict[str, float]) -> None:
    fig, axes = plt.subplots(figsize=(9.5, 3.2), ncols=3)
    fig.suptitle("Distributions of $s_{12}$, $s_{23}$, and $s_{31}$")
    for ax, ij in zip(axes, s, strict=True):
        ax.hist(s[ij], bins=100, color="black", histtype="step")
        ax.set_xlabel(axis_label(ij))
        ax.set_ylabel("counts")
        ax.axvline(peaks[ij], c=COLORS[ij], ls="dashed", label=f"$R_{{{ij}}}$")
        ax.legend()
    fig.tight_layout()
    plt.show()


def plot_bands(
    s: dict[str, np.ndarray], peaks: dict[str, float], width: float = 0.08
) -> None:
    fig, axes = plt.subplots(figsize=(9.5, 3.2), ncols=3)
    fig.suptitle("Event distributions along the resonance bands")
    for ax, (band, other) in zip(
        axes, [("12", "23"), ("23", "31"), ("31", "12")], strict=True
    ):
        selection = np.abs(s[band] - peaks[band]) < width
        ax.hist(s[other][selection], bins=60, color="black", histtype="step")
        ax.set_title(Rf"$|s_{{{band}}} - R_{{{band}}}| < {width}\text{{ GeV}}^2$")
        ax.set_xlabel(axis_label(other))
        ax.set_ylabel("counts")
    fig.tight_layout()
    plt.show()
````

````python
s = {
    "12": mass_squared(p1 + p2),
    "23": mass_squared(p2 + p3),
    "31": mass_squared(p3 + p1),
}
peaks = {ij: find_peak(s_ij) for ij, s_ij in s.items()}
{ij: round(float(np.sqrt(r)), 3) for ij, r in peaks.items()}
````

````python
plot_dalitz(s, peaks)
````

````python
plot_projections(s, peaks)
````

The particle candidates for $R_{12} \to \eta\pi^0$ and $R_{23} \to \pi^0 p$ are:

````python
find_candidates(np.sqrt(peaks["12"]), delta=0.01, charge=0)
````

````python
find_candidates(np.sqrt(peaks["23"]), delta=0.01, charge=+1)
````

The $\Xi^0$ baryon cannot decay into two mesons, so the $\eta\pi^0$ resonance is the $a_2(1320)$. Likewise, the $b_1(1235)^+$ and $a_1(1260)^+$ mesons and the anti-baryon $\bar\Delta(1232)^+$ cannot decay into $\pi^0 p$, so the $\pi^0 p$ resonance is the $\Delta(1232)^+$.

The $p\eta$ band is wider. Since the $\eta$ has isospin&nbsp;0, a $p\eta$ resonance has isospin&nbsp;1/2, which excludes the $\Delta$ states. The remaining candidates within 50&nbsp;MeV are:

````python
[
    p
    for p in find_candidates(np.sqrt(peaks["31"]), delta=0.05, charge=+1)
    if p.pdgid > 0 and np.isclose(p.I, 1 / 2)
]
````

Of these, the $N(1535)^+$ is known for its strong coupling to $N\eta$, so it is the most likely candidate, even though the peak lies somewhat above its nominal mass. In summary, the reaction is $\gamma p \to \eta\pi^0 p$ with the resonances $a_2(1320) \to \eta\pi^0$, $\Delta(1232)^+ \to \pi^0 p$, and $N(1535)^+ \to p\eta$.

:label: three-particles-1-spin
What can you say about the spin of the resonances?
:::

:::{solution-start} three-particles-1-spin
:class: dropdown full-width
:::

A resonance with spin $J$ generally produces an angular distribution of its decay products in its rest frame. Along a resonance band in the Dalitz plot, the other invariant masses are linear functions of the cosine of this decay angle, so an angular distribution shows up as a modulation of the event density along the band. We therefore plot the distribution of events within each band.

````python
plot_bands(s, peaks)
````

Apart from the peaks at the edges, where the bands cross the other resonance bands, the event density along each band is roughly flat. These data therefore show no sign of the spin of the resonances, even though the $a_2(1320)$ has $J=2$ and the $\Delta(1232)$ has $J=3/2$: the sample was apparently generated without an angular dependence.
