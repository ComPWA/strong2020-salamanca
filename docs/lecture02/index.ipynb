## Experiments and frames

Particle kinematics is the application of special relativity to elementary particle reactions.
In experiments, the "beam" collides onto the "target" with high energy to create other particles.

Each particle has a 4-vector $p^{\mu}$:

$$
p^{\mu} = (E, p_x, p_y, p_z) = (E, \vec{p})
$$

Two frames that are commonly used are the Center of Mass (CoM) frame and the laboratory (Lab) frame.

For a two-particle reaction, $a+b \to 1 + ... + n$, the frames satisfy the following relations.
- **Lab frame**: $\vec{p}_b = \vec{0}$<br>
  Target particle $b$ is at rest, while beam particle $a$ is moving.
- **CoM frame**: $\vec{p}_a+\vec{p}_b = \vec{0}$<br>
  The total momentum is zero, i.e., the momenta of the two initial state particles have opposite direction.

Masses can be defined as a Lorentz invariant quantity (unaffected by reference frame):

$$
p^2_i = E^2_i - |\vec{p}_i|^2 = m^2_i,
$$

where $p$ is the 4-vector, $E$ is the energy, $\vec{p}$ is the three vector, and $m$ is the mass.

When we have a final state with two particles ($n=2$), we can define one $xz$ plane that is spanned by the momenta of the two final state particles in the Lab frame. The CoM frame and the Lab frame can now be related by a boost along the $z$ axis (the direction of particle $a$).

With this definition of the $xz$ plane, we have:
- $\vec{z}$ is parallel to the beam $\vec{p}_a$
- $\vec{y}$ is parallel to $\vec{p}_a \times \vec{p}_1$

## Lorentz transformations

Lorentz transformations can be factorized into rotations and boosts along one axis and are required for transforming particle properties, such as position and time, between different inertial frames in relative motion.

### Boost along one axis

When we boost in the direction of the $z$ axis, only the energy and the $z$ component of the four-momenta changes, so we can express the Lorentz boost as a 2D matrix:

$$\begin{pmatrix}
 E^*
 \\
 p_z^*
\end{pmatrix} =
\begin{pmatrix}
  \gamma & \gamma \beta \\
  \gamma \beta & \gamma
\end{pmatrix}
\begin{pmatrix}
 E^L\\
 p^L_z
\end{pmatrix},
$$

and its inverse relation,

$$\begin{pmatrix}
 E^L
 \\
 p_z^L
\end{pmatrix} =
\begin{pmatrix}
  \gamma & -\gamma \beta \\
  -\gamma \beta & \gamma
\end{pmatrix}
\begin{pmatrix}
 E^*\\
 p^*_z
\end{pmatrix},
$$ (inverse_relation)

with

$$
\beta = \frac{p^*_{b,z}}{E^*_b}
$$ (beta_label)

and

$$
\gamma = \frac{1}{\sqrt{1-\beta^2}} = \frac{E^*_b}{m_b},
$$ (gamma_label)

:::

:::{solution} boost-exercise
:class: dropdown

We want to show that the boost {eq}`inverse_relation` from the CoM frame to the lab frame brings the target $b$ to rest, that is,

$$
\begin{pmatrix}
 E_b^L \\
 p_{b,z}^L
\end{pmatrix} =
\begin{pmatrix}
 m_b \\
 0
\end{pmatrix}.
$$

In the CoM frame, the target moves along the $z$ axis, so $p_{b,x}^* = p_{b,y}^* = 0$. A boost along the $z$ axis does not change these components, so they also vanish in the lab frame and we only have to consider the energy and the $z$ component.

We first express $\gamma$ and $\gamma\beta$ in terms of the CoM energy and momentum of the target. Inserting the velocity {eq}`beta_label` into the definition of $\gamma$ and using the mass-shell relation $E_b^{*2} - p_{b,z}^{*2} = m_b^2$ gives

$$
\gamma = \frac{1}{\sqrt{1-\beta^2}} = \frac{E_b^*}{\sqrt{E_b^{*2} - p_{b,z}^{*2}}} = \frac{E_b^*}{m_b},
\qquad
\gamma\beta = \frac{E_b^*}{m_b} \cdot \frac{p_{b,z}^*}{E_b^*} = \frac{p_{b,z}^*}{m_b},
$$

which is the result in {eq}`gamma_label`. Substituting these into {eq}`inverse_relation` for particle $b$, we get

$$
\begin{pmatrix}
 E_b^L \\
 p_{b,z}^L
\end{pmatrix} =
\frac{1}{m_b}
\begin{pmatrix}
  E_b^* & -p_{b,z}^* \\
  -p_{b,z}^* & E_b^*
\end{pmatrix}
\begin{pmatrix}
 E_b^* \\
 p_{b,z}^*
\end{pmatrix} =
\frac{1}{m_b}
\begin{pmatrix}
 E_b^{*2} - p_{b,z}^{*2} \\
 -p_{b,z}^* E_b^* + E_b^* p_{b,z}^*
\end{pmatrix} =
\begin{pmatrix}
 m_b \\
 0
\end{pmatrix}.
$$

The target therefore has no momentum in the lab frame and its energy equals its mass, so it is at rest.
:::

### Rotations

We use _active_ rotations, meaning that the momentum itself is rotated while the coordinate axes remain fixed.

As an example, take a momentum of unit length in the $xz$ plane that forms an angle $\theta$ with the $z$ axis:

$$
\begin{pmatrix}
 p_x
 \\
 p_y
 \\
 p_z
\end{pmatrix} =
\begin{pmatrix}
 \sin(\theta)
 \\
 0
 \\
 \cos(\theta)
\end{pmatrix}.
$$

After a rotation over an angle $\omega$ around the $y$ axis, the rotated momentum forms an angle $\theta + \omega$ with the $z$ axis:

$$
\begin{pmatrix}
 p_x'
 \\
 p_y'
 \\
 p_z'
\end{pmatrix}
=
R_y(\omega)
\begin{pmatrix}
 p_x
 \\
 p_y
 \\
 p_z
\end{pmatrix}
=
\begin{pmatrix}
 \sin(\theta+\omega)
 \\
 0
 \\
 \cos(\theta+\omega)
\end{pmatrix}.
$$ (rotation_label)

:::{solution} check-rotation

Recall the rotation matrices. Any rotation can be decomposed into rotations around the $z$ and $y$ axes:

$$
R_z(\omega)
=
\begin{pmatrix}
\cos(\omega) & -\sin(\omega) & 0 \\
\sin(\omega) & \cos(\omega) & 0 \\
0 & 0 & 1
\end{pmatrix},
$$

$$
R_y(\omega)
=
\begin{pmatrix}
\cos(\omega) & 0 & \sin(\omega) \\
0 & 1 & 0 \\
-\sin(\omega) &0 & \cos(\omega)
\end{pmatrix}.
$$

The rotation matrix around the $x$ axis is also given for completeness:

$$
R_x(\omega)
=
\begin{pmatrix}
1&0&0 \\
0 & \cos(\omega) & -\sin(\omega) \\
0 & \sin(\omega) & \cos(\omega)
\end{pmatrix}.
$$

With this, we can rewrite the rotation of {eq}`rotation_label` as

$$
\begin{pmatrix}
 p_x' \\
 p_y' \\
 p_z'
\end{pmatrix}
=
R_y(\omega)
\begin{pmatrix}
 p_x \\
 p_y \\
 p_z
\end{pmatrix}
=
\begin{pmatrix}
\cos(\omega) & 0 & \sin(\omega) \\
0 & 1 & 0 \\
-\sin(\omega) &0 & \cos(\omega)
\end{pmatrix}
\begin{pmatrix}
 p_x \\
 p_y \\
 p_z
\end{pmatrix}
=
\begin{pmatrix}
p_x \cdot \cos(\omega)+ p_z \cdot \sin(\omega) \\
p_y \\
-p_x \cdot \sin(\omega) + p_z \cdot \cos(\omega)
\end{pmatrix}.
$$

Since $(p_x, p_y, p_z) = (\sin(\theta), 0, \cos(\theta))$, and using the trigonometric identities $\sin(\theta+\omega) = \sin(\theta)\cos(\omega) + \cos(\theta)\sin(\omega)$ and $\cos(\theta+\omega) = \cos(\theta)\cos(\omega) - \sin(\theta)\sin(\omega)$,
we get:

$$
\begin{pmatrix}
 p_x' \\
 p_y' \\
 p_z'
\end{pmatrix}
=
\begin{pmatrix}
\sin(\theta) \cdot \cos(\omega)+ \cos(\theta) \cdot \sin(\omega) \\
0 \\
-\sin(\theta) \cdot \sin(\omega) + \cos(\theta) \cdot \cos(\omega)
\end{pmatrix}
=
\begin{pmatrix}
\sin(\theta+\omega) \\
0 \\
\cos(\theta+\omega)
\end{pmatrix}.
$$
:::

The data file lists the four-momenta $p_a$, $p_1$, and $p_2$ of the reaction $a+b \to 1+2$ for each event, one after the other. We split them into three arrays.

````python
pa = data[::3].T
p1 = data[1::3].T
p2 = data[2::3].T
pa.shape
````

The four-momentum of the target $p_b$ follows from four-momentum conservation, $p_a + p_b = p_1 + p_2$. We check that it is the same for all events.

````python
pb = p1 + p2 - pa
assert pb.std(axis=1).sum().round(5) == 0
pb.mean(axis=1).round(3)
````

The mass of particle&nbsp;1 is given by $m_1 = \sqrt{E_1^2 - |\vec{p}_1|^2}$:

````python
m1 = np.sqrt(p1[0] ** 2 - p1[1] ** 2 - p1[2] ** 2 - p1[3] ** 2)
m1.mean(), m1.std()
````

and the length of its three-momentum is:

````python
np.sqrt((p1[1:] ** 2).sum(axis=0))
````

Since the reaction takes place in the $xz$ plane, we can compute $\cos\theta$ with

$$
\cos \theta = \frac{p_{1z}}{|\vec{p_1}|}
$$(cos_theta_label)

````python
def p3_length(four_momentum: np.ndarray) -> np.ndarray:
    return np.sqrt(
        four_momentum[1] ** 2 + four_momentum[2] ** 2 + four_momentum[3] ** 2
    )


p1_length = p3_length(p1)
cos_theta = p1[3] / p1_length
````

````python
p1_length_vectorized = np.sqrt(np.sum(p1[1:] ** 2, axis=0))
np.testing.assert_array_equal(p1_length, p1_length_vectorized)
````

#### $\cos\theta$ in the CoM frame

:::{exercise}
:label: two-particles-cm
Boost the four-momenta of {ref}`two-particles-lab` to the CoM frame and compute and plot the $\cos\theta^*$ distribution.
:::

:::{solution-start} two-particles-cm
:class: dropdown

We first perform the boost by hand with NumPy.

The CoM frame moves with velocity $\vec{\beta} = \vec{P}/E$ with respect to the lab frame, where $P = p_a + p_b = (E, \vec{P})$ is the total four-momentum of the initial state. Since the target is at rest in the lab frame, $\vec{P} = \vec{p}_a$ and $E = E_a + m_b$.

````python
P = pa + pb
β_CM = P[1:] / P[0]
β2 = np.sum(β_CM**2, axis=0)
gamma = 1 / np.sqrt(1 - β2)
````

The energy of particle&nbsp;1 in the CoM frame follows from the Lorentz boost

$$
E_1^* = \gamma\left(E_1 - \vec{\beta} \cdot \vec{p}_1\right),
$$

with $\gamma = 1/\sqrt{1-\beta^2}$.

````python
β_dot_p1 = np.sum(β_CM * p1[1:], axis=0)
E1_CM = gamma * (p1[0] - β_dot_p1)
````

Its three-momentum in the CoM frame is

$$
\vec{p}_1^{\,*} = \vec{p}_1 + \left(\frac{\gamma - 1}{\beta^2} \left(\vec{\beta} \cdot \vec{p}_1\right) - \gamma E_1\right) \vec{\beta}.
$$

````python
p1_CM_vec = p1[1:] + ((gamma - 1) / β2 * β_dot_p1 - gamma * p1[0]) * β_CM
p1_CM = np.concatenate((E1_CM[None], p1_CM_vec))
````

Equation {eq}`cos_theta_label` also holds in the CoM frame, so we find $\cos\theta^*$ with:

````python
p1_CM_length = p3_length(p1_CM)
cos_theta_CM = p1_CM[3] / p1_CM_length
````

````python
plt.hist(cos_theta_CM, bins=40)
plt.title(r"$\cos \theta$ distribution in CoM frame")
plt.xlabel(r"$\cos \theta$")
plt.show()
````

We check the boost with the [`pylorentz`](https://pypi.org/project/pylorentz) package and compare the $\cos\theta^*$ values computed in both ways. The method `boost_particle()` transforms a four-vector from the rest frame of the given momentum into the lab frame. To transform from the lab frame into the CoM frame, we therefore pass it the total four-momentum with reversed three-momentum.

````python
from pylorentz import Momentum4

p1_Lab = Momentum4(*p1)
pa_Lab = Momentum4(*pa)
pb_Lab = Momentum4(*pb)

cm = pa_Lab + pb_Lab
cm_reversed = Momentum4(cm.e, -cm.p_x, -cm.p_y, -cm.p_z)
p1_CM_pyL = p1_Lab.boost_particle(cm_reversed)

cos_theta_CM_pyL = p1_CM_pyL.p_z / p1_CM_pyL.p

np.testing.assert_allclose(cos_theta_CM, cos_theta_CM_pyL, rtol=1e-5)
print("Test passed with relative tolerance 1e-5")
````

````python
plt.hist(cos_theta_CM_pyL, bins=40)
plt.title(r"$\cos \theta$ distribution in CoM frame")
plt.xlabel(r"$\cos \theta$")
plt.show()
````

## Mandelstam variables

We have the mass-shell relation:

$$ p_i^2 = m^2_i $$

The Mandelstam variables s, t, and u are Lorentz invariants. Their definition are:

$$ s =(p_a + p_b)^2 = (p_1+p_2)^2 $$
$$ t =(p_a - p_1)^2 = (p_b-p_2)^2 $$
$$ u =(p_a - p_2)^2 = (p_b-p_1)^2 $$

```{image} https://github.com/ComPWA/strong2020-salamanca/assets/17490173/4daa6f59-615d-4950-b7ba-e8a93f04253c
:width: 300px
```

$$
s+t+u = m_a^2+m_b^2+m_1^2+m^2_2
$$ (label:mv)

:::{solution} mv

$$ s+t+u = (p_a + p_b)^2 + (p_a - p_1)^2 + (p_a - p_2)^2$$

$$ s+t+u = p_a^2 +p_b^2 + 2 p_a p_b + p_a^2 -2 p_a p_1 + p_1^2 + p_a^2 - 2p_a p_2 +p_2^2$$

$$
s+t+u = 3p_a^2 + p_b^2 +p_1^2 + p_2^2 + 2p_a p_b - 2p_a p_1 -2 p_a p_2
$$ (label:proving)

from equation {eq}`label:mv` we have:

$$ p_b = p_1 + p_2 - p_a $$

$$
p_a \cdot p_b = p_a \cdot p_1 + p_a \cdot p_2 - p_a^2
$$ (label:use)

By equation {eq}`label:proving` and substitute it into equation {eq}`label:use`:

$$
s+t+u = 3p_a^2 + p_b^2 +p_1^2 + p_2^2 + 2p_a \cdot p_1 + 2p_a \cdot p_2 - 2p_a^2 -2p_a p_1 -2 p_a p_2
$$

$$
s+t+u = p_a^2 + p_b^2 +p_1^2 + p_2^2
$$

by the mass shell relation, and thus:

$$s+t+u = m_a^2 + m_b^2 + m_1^2 + m_2^2$$

````python
n_final_state = 3
pa, p1, p2, p3 = (data[i::4].T for i in range(n_final_state + 1))
p0 = p1 + p2 + p3
pb = p0 - pa
````

````python
def mass(p: np.ndarray) -> np.ndarray:
    return np.sqrt(mass_squared(p))


def mass_squared(p: np.ndarray) -> np.ndarray:
    return p[0] ** 2 - np.sum(p[1:] ** 2, axis=0)


m0 = mass(p0)
print(f"{m0.mean():.4g} +/- {m0.std():.4g}")
````

````python
from IPython.display import Math

display(Math(Rf"m_a = {mass(pa).mean():.3g}\text{{ GeV}}"))
display(Math(Rf"m_b = {mass(pb).mean():.3g}\text{{ GeV}}"))
for i, p in enumerate([p0, p1, p2, p3]):
    display(Math(Rf"m_{i} = {mass(p).mean():.3g}\text{{ GeV}}"))
````

````python
from particle import Particle


def find_candidates(
    mass: float, delta: float = 0.001, charge: float | None = None
) -> list[Particle]:
    def identify(p) -> bool:
        if p.pdgid == 21:
            return None
        if charge is not None and p.charge != charge:
            return False
        return mass - delta < 0.001 * p.mass < mass + delta

    return Particle.findall(identify)


ma = mass(pa).mean()
mb = mass(pb).mean()
m1 = mass(p1).mean()
m2 = mass(p2).mean()
m3 = mass(p3).mean()
initial_state = (
    find_candidates(ma.mean(), delta=1e-4)[0],
    find_candidates(mb.mean())[0],
)
final_state = tuple(find_candidates(m.mean())[0] for m in [m1, m2, m3])
display(
    Math(R"\text{Incoming: }" + ", ".join(f"{p.latex_name}" for p in initial_state)),
    Math(R"\text{Outgoing: }" + ", ".join(f"{p.latex_name}" for p in final_state)),
)
````

So this is a photon&nbsp;$\gamma$ hitting a proton&nbsp;$p$ and producing a meson&nbsp;$\eta$, pion&nbsp;$\pi^0$, and proton&nbsp;$p$.
